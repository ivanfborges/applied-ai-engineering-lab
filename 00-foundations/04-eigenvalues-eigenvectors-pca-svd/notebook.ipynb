# Eigenvectors, PCA, and SVD Visual Laboratory

This notebook is a lightweight entrypoint into the reusable `visual_lab` package. All data is synthetic and reproducible.

```python
import numpy as np
from visual_lab.datasets import correlated_2d, correlated_3d
from visual_lab.math_utils import eigendecomposition_pca, svd_pca
from visual_lab.plotting import covariance_pca_figure, pca_3d_figure
```

## Covariance eigenvectors become principal directions

The arrows have lengths proportional to the standard deviation along each component, $\sqrt{\lambda_i}$.

```python
data_2d = correlated_2d(seed=42)
covariance_pca_figure(data_2d)
```

## PCA and SVD equivalence

For centered data, $\lambda_i = \sigma_i^2/(n-1)$. Component signs may differ without changing the subspace.

```python
data_3d = correlated_3d(seed=7)
eigen_result = eigendecomposition_pca(data_3d, n_components=3)
svd_result = svd_pca(data_3d, n_components=3)
print(np.column_stack([eigen_result.explained_variance, svd_result.explained_variance]))
pca_3d_figure(data_3d).show()
```
